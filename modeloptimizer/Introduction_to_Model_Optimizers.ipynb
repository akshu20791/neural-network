# Introduction to Model Optimizers

## What is an Optimizer?
An optimizer is an algorithm that helps a neural network reduce its error (loss) by updating weights and biases during training.

## Real-Life Analogy
Imagine you are climbing down a mountain in fog. You cannot see the bottom, so you take small steps in the direction where the ground slopes downward. Optimizers work similarly to find the minimum loss.

## Why Do We Need Optimizers?
- Reduce model error
- Improve accuracy
- Speed up training
- Help models converge efficiently

## Popular Optimizers
1. Gradient Descent
2. Stochastic Gradient Descent (SGD)
3. Momentum
4. RMSProp
5. Adam

```python
# Simple optimizer demonstration
weight = 10
learning_rate = 0.1

for epoch in range(5):
    gradient = 2 * weight
    weight = weight - learning_rate * gradient
    print(f'Epoch {epoch+1}: Weight = {weight}')
```
