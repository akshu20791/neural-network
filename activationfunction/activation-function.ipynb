```python
# pass the same values through different activation fucntions and compare the outputs
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
#create input values
x = np.arange(-10,11,1) 
print(x)
```

```python
#sigmoid function
def sigmoid(x):
    return 1/(1 + np.exp(-x))
y = sigmoid(x)
print(y)
```

```python
plt.plot(x,y)
plt.title("sigmoid")
plt.grid()
plt.show()
```

```python
#tanh activation:
def tanh(x):
    return np.tanh(x)
y = tanh(x)
plt.plot(x,y)
plt.title("tanh")
plt.grid()
plt.show()
```

```python
#RELU activation

def relu(x):
    return np.maximum(0,x)
y = relu(x)
plt.plot(x,y)
plt.title("relu")
plt.grid()
plt.show()
```

```python
#softmax
scores = np.array([2.5, 1.2, 0.3])
softmax = np.exp(scores) / np.sum(np.exp(scores))
```
